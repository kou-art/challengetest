```python
import os
import importlib.metadata
from pathlib import Path

# pip版 libclang の実体を探す
dist = importlib.metadata.distribution("libclang")

libclang_path = None

for file in dist.files:
    name = str(file)

    if name.endswith("libclang.so"):
        libclang_path = Path(dist.locate_file(file))
        break

if libclang_path is None:
    raise RuntimeError("libclang.so が見つかりません")

print("libclang:")
print(libclang_path)
```

```python
from pathlib import Path
from clang import cindex
from clang.cindex import CursorKind
import os
import re


# ============================================================
# 1. 入力
# ============================================================

PROJECT_ROOT = Path(input("Cプロジェクトのルートフォルダ: ").strip()).resolve()

TARGET_VARIABLE = input("調べたい変数名: ").strip()

if not PROJECT_ROOT.exists():
    raise FileNotFoundError(
        f"フォルダが存在しません: {PROJECT_ROOT}"
    )

if not PROJECT_ROOT.is_dir():
    raise NotADirectoryError(
        f"フォルダではありません: {PROJECT_ROOT}"
    )

if not TARGET_VARIABLE:
    raise ValueError("変数名が空です。")


# ============================================================
# 2. .c / .h を再帰探索
# ============================================================

c_files = sorted(
    p for p in PROJECT_ROOT.rglob("*")
    if p.is_file() and p.suffix.lower() == ".c"
)

h_files = sorted(
    p for p in PROJECT_ROOT.rglob("*")
    if p.is_file() and p.suffix.lower() in {".h", ".hpp"}
)

print()
print("============================================================")
print("Project")
print("============================================================")
print(f"Root : {PROJECT_ROOT}")
print(f".c   : {len(c_files)}")
print(f".h   : {len(h_files)}")


if not c_files:
    raise RuntimeError(
        f".cファイルが見つかりません: {PROJECT_ROOT}"
    )


# ============================================================
# 3. プロジェクト内のincludeディレクトリを自動生成
#
# ここではフォルダ名を決め打ちしない。
# すべての .c/.h が存在するディレクトリを候補にする。
#
# ※ 本番では compile_commands.json がある場合、
#    それを使う方が正確。
# ============================================================

include_dirs = sorted(
    {
        str(p.parent)
        for p in c_files + h_files
    }
)

clang_args = [
    "-x", "c",
    "-std=c11",
]

for include_dir in include_dirs:
    clang_args.append(f"-I{include_dir}")


# ============================================================
# 4. Clang index
#
# index は libclang 初期化済みである前提。
# ============================================================

index = cindex.Index.create()


# ============================================================
# 5. .c を全部解析
# ============================================================

translation_units = []

print()
print("============================================================")
print("Clang parsing")
print("============================================================")

for c_file in c_files:

    try:

        tu = index.parse(
            str(c_file),
            args=clang_args,
            options=0
        )

        # 致命的エラー
        errors = [
            d for d in tu.diagnostics
            if d.severity >= 3
        ]

        if errors:

            print(f"[ERROR] {c_file}")

            for d in errors:
                print(f"  {d}")

        else:

            translation_units.append(tu)
            print(f"[OK] {c_file}")

    except Exception as e:

        print(f"[EXCEPTION] {c_file}")
        print(f"  {e}")


print()
print(
    f"解析成功: "
    f"{len(translation_units)} / {len(c_files)}"
)

if not translation_units:
    raise RuntimeError(
        "解析に成功したCファイルがありません。"
    )


# ============================================================
# 6. 対象変数の宣言・定義を取得
#
# spelling完全一致。
#
# g を指定した場合、
# g_temperature や global_g は一致させない。
# ============================================================

target_declarations = []


def collect_declarations(cursor):

    if cursor.kind == CursorKind.VAR_DECL:

        if cursor.spelling == TARGET_VARIABLE:

            target_declarations.append(cursor)

    for child in cursor.get_children():
        collect_declarations(child)


for tu in translation_units:
    collect_declarations(tu.cursor)


# ============================================================
# 7. 同一変数のUSRを取得
#
# extern宣言と実体定義が同じ変数なら、
# Clangが同一USRとして扱えるものをまとめる。
# ============================================================

target_usrs = set()

for cursor in target_declarations:

    usr = cursor.get_usr()

    if usr:
        target_usrs.add(usr)


# ============================================================
# 8. 対象変数を使用している関数を取得
# ============================================================

used_functions = {}


def walk(cursor, current_function=None):

    # --------------------------------------------------------
    # 関数定義
    # --------------------------------------------------------

    if cursor.kind == CursorKind.FUNCTION_DECL:

        if cursor.is_definition():
            current_function = cursor


    # --------------------------------------------------------
    # 変数参照
    # --------------------------------------------------------

    if cursor.kind == CursorKind.DECL_REF_EXPR:

        # 名前をまず完全一致
        if cursor.spelling == TARGET_VARIABLE:

            referenced = cursor.referenced

            if referenced is not None:

                referenced_usr = referenced.get_usr()

                # 対象変数の宣言/定義を参照しているか
                if referenced_usr in target_usrs:

                    if current_function is not None:

                        function_usr = current_function.get_usr()

                        if function_usr:

                            file_name = None

                            if current_function.location.file:
                                file_name = (
                                    current_function
                                    .location
                                    .file
                                    .name
                                )

                            used_functions[function_usr] = {
                                "name":
                                    current_function.spelling,

                                "file":
                                    file_name,

                                "line":
                                    current_function.location.line,
                            }


    # --------------------------------------------------------
    # 子AST
    # --------------------------------------------------------

    for child in cursor.get_children():
        walk(child, current_function)


for tu in translation_units:
    walk(tu.cursor)


# ============================================================
# 9. 結果
# ============================================================

print()
print("============================================================")
print(f"対象変数: {TARGET_VARIABLE}")
print("============================================================")

if not target_declarations:

    print("対象変数の宣言・定義が見つかりません。")

else:

    print()
    print("宣言・定義:")

    declaration_keys = set()

    for cursor in target_declarations:

        if cursor.location.file is None:
            continue

        file_name = cursor.location.file.name
        line = cursor.location.line
        key = (file_name, line)

        if key in declaration_keys:
            continue

        declaration_keys.add(key)

        print(
            f"  {cursor.spelling}"
            f" : {cursor.type.spelling}"
            f" : {file_name}:{line}"
        )


print()
print("============================================================")
print(f"「{TARGET_VARIABLE}」を使用している関数")
print("============================================================")

if not used_functions:

    print("該当する関数は見つかりませんでした。")

else:

    results = sorted(
        used_functions.values(),
        key=lambda x: (
            x["file"] or "",
            x["line"] or 0,
            x["name"]
        )
    )

    for i, func in enumerate(results, 1):

        print(
            f"{i}. {func['name']}"
            f"    {func['file']}:{func['line']}"
        )

    print()
    print(f"合計: {len(results)} 関数")
```
